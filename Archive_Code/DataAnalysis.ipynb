Stats.json Data

```python
import json
```

```python
with open("../Stats.json", "r", encoding="utf-8") as f:
    stats = json.load(f)

print(f"Total Count: {stats['count']:,}")
print(f"Different Users: {stats['different_users']:,}\n")

# Sort by count
top10 = sorted(
    stats['cities'].items(),
    key=lambda x : x[1],
    reverse=True
)[:10]

print("Overall Top 10 Cities:\n")
for city, count in top10:
    print(f"{city}: {count:,}")

print("")

month_key = ["January", "February", "March", "April", "May", "June", "July", "August", "September", "October", "November", "December"]

for month, data in stats["monthly"].items():

    month_name = month_key[int(month) - 1]

    print(f"--- {month_name} ---\n")

    print(f"Total Tweets: {data['count']:,}")
    print(f"Total Different Users: {data['different_users']:,}\n")

    print(f"Top 10 Cities in {month_name}")

    m_top10 = sorted(
        data['cities'].items(),
        key=lambda x: x[1],
        reverse=True
    )[:10]

    for city, count in m_top10:
        print(f"{city}: {count:,}")

    print()
```

San Fransisco Map

```python
import geopandas as gpd
import folium

gdf = gpd.read_file("Shapefiles/California/tl_2017_06_tract.shp")
gdf = gdf.to_crs(epsg=4326)

m = folium.Map(
    location=[37.77, -122.43],
    zoom_start=12,
    tiles="cartodbpositron"
)

tooltip = folium.GeoJsonTooltip(
    fields=["TRACTCE"],          # or "GEOID"
    aliases=["Tract:"],
    localize=True,
    sticky=False,
    labels=True,
    style="""
        background-color: white;
        border: 1px solid black;
        border-radius: 3px;
        box-shadow: 3px;
    """
)

folium.GeoJson(
    gdf,
    name="Census Tracts",
    style_function=lambda x: {
        "fillColor": "#3388ff",
        "color": "black",
        "weight": 0.5,
        "fillOpacity": 0.4
    },
    tooltip=tooltip
).add_to(m)

m.save("sf_census_tracts.html")
gdf.head()
```
